# Task 1

In this code, I wrote my own humorous Saaremaa-themed text and stored it in a variable called sentence. I printed the original text, then split it into a list of words using .split(), which allowed me to work with each word individually. After splitting the text, I performed several operations: I printed the first and last five words, selected every third word, found all long words (words with six or more letters), and created new lists using list comprehensions. I also generated a small “secret code” made from the first letter of each word.


Saaremaa-themed text for processing

```python
sentence = """Saaremaa is a windy place, and everyone on Saaremaa talks about the wind. 
The wind in Saaremaa is so strong that even the sheep complain about the wind, and the sheep of Saaremaa complain loudly. 
Some sheep claim the wind once stole their hats, but no one knows why sheep wear hats on Saaremaa anyway. 
On windy mornings, people drink hot coffee, watch the sheep fight the wind, and laugh as the sheep slide across the muddy fields. 
Saaremaa mud is famous too, very sticky, very slippery, and very good at stealing boots. 
Visitors say Saaremaa mud is actually a secret pirate weapon because Saaremaa pirates love trapping tourists in mud while they eat pancakes. 
Pancakes are important on Saaremaa. Windy days need pancakes. Muddy days need pancakes. Even windmill-opening ceremonies need pancakes. 
The old windmills of Saaremaa turn wildly in the wind, almost dancing, while the sheep cheer and the pirates steal more pancakes. 
In the evenings, everyone on Saaremaa: people, sheep, pirates, and even the windmills, gathers to eat pancakes and talk about the wind, the mud, and the mysteries of Saaremaa.
"""
```

Print the original text

```python
print("ORIGINAL TEXT:")
print(sentence)
print("\n")
```

Split into words

```python
words = sentence.split()
print("Words list:")
print(words)
print("\nTotal words:", len(words))
print("\n")
```

Print the first 5 words

```python
print("First 5 words:", words[:5])
```

Print the last 5 words

```python
print("Last 5 words:", words[-5:])
```

Print every 3rd word

```python
print("Every 3rd word:", words[::3])
```

Words that are 6 letters or longer

```python
long_words = [w for w in words if len(w) >= 6]
print("Long words (6+ letters):", long_words)
```

First and last letter of each word

```python
first_last = [word[0] + word[-1] for word in words]
print("First+last letters of each word:", first_last)
```

Join the first letters to create a 'secret code'

```python
secret_code = "".join([word[0] for word in words])
print("Secret code from first letters:", secret_code)
```

```python
import urllib.request
from wordcloud import WordCloud
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1) Load TEXT from Drive
text_url = "https://drive.google.com/uc?export=download&id=1jrTsGXSoe1Tzc5k3YQiVVeHbbDG15pVa"
text = urllib.request.urlopen(text_url).read().decode("utf-8")

# 2) Load MASK PNG from Drive
mask_url = "https://drive.google.com/uc?export=download&id=1vTRy-f1jHbCKsZy9QuyTjlQnWL5oiI9X"
urllib.request.urlretrieve(mask_url, "saaremaa_shape.png")

# 3) Prepare mask (ONLY ONCE)
mask_img = Image.open("saaremaa_shape.png").convert("L")
mask = np.array(mask_img)

# dark = shape (block), light = background (allow words)
mask = np.where(mask < 200, 0, 255).astype(np.uint8)

# If words appear outside shape, uncomment:
# mask = 255 - mask

# 4) Custom color function (Exercise 18)
def color_function(word, font_size, position, orientation, random_state=None, **kwargs):
    if len(word) < 5:
        return "green"
    return "blue"

# 5) Generate word cloud
wc = WordCloud(
    mask=mask,
    background_color="white",
    stopwords=["and", "the", "on", "in", "is", "a", "of", "to", "as", "but"],
    collocations=False
).generate_from_text(text).recolor(color_func=color_function)

# 6) Show result clearly in notebook
plt.figure(figsize=(12,6))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```
