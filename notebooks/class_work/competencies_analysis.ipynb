```python
import pandas as pd
df = pd.read_csv('competencies.csv')
df
```

Who puts in the most effort?

```python
import matplotlib.pyplot as plt

avg_hours = df.groupby('Student')['Hours_Studied'].mean().reset_index()
#Bar chart of study hours per student
avg_hours = avg_hours.plot(kind='bar', x='Student', y='Hours_Studied', title='Study Hours per Student')

#Titles and labels
plt.title('Study Hours per Student')
plt.xlabel('Student')
plt.ylabel('Hours Studied')
plt.show()
```

```python
# Average study hours per student
avg_grades = df.groupby('Student')['Grade'].mean().sort_values(ascending=False)

# Bar chart of study hours per student
avg_grades.plot(kind='bar', x='Student', y='Grade', legend=False)

plt.title("Who performs the best? (by analysing average grades per student)")
plt.ylabel('Grade')

plt.show()
```

```python
# Average study hours per student
avg_grades = df.groupby('Student')['Grade'].mean().sort_values(ascending=False)

plt.figure(figsize=(8,8)) # Set the figure size to be 8x8 inches

#plotting a pie chart
plt.pie(avg_grades, labels=avg_grades.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)

#colors=plt.cm.Paired.colors= a color map to get different colors for each slice
#startangle=140= rotates the start of the pie chart by 140 degrees
#autopct='%1.1f%%'= to show the percentage value on each slice

plt.show()
```

```python
#Does more studying = better grades?

plt.scatter(df['Hours_Studied'], df['Grade'], s=df['Hours_Studied']*30, alpha=0.5)
plt.title('Study Hours vs Grade')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')
plt.title("Grades vs hours studied (bubbles sized by hours studied)")
plt.show()
```

```python
#Which course has the highest average student performance
avg_course = df.groupby('Course')['Grade'].mean()
print("Average grade per course:\n", avg_course, "\n")

avg_course.plot(kind='bar', figsize=(8,6), color='skyblue')
plt.title('Average Grade per Course')
plt.ylabel('Average Grade')
plt.show()

#which program has the highest average student performance
avg_program = df.groupby('Program')['Grade'].mean()
print("Average grade per program:\n", avg_program, "\n")

avg_program.plot(kind='bar', figsize=(6,4), color='lightgreen')
plt.title('Average Grade per Program')
plt.ylabel('Average Grade')
plt.show()
```

```python
plt.figure(figsize=(8,5)) # Set the figure size to be 8x5 inches
df[["Hours_Studied", "Grade"]].boxplot()
plt.title('Boxplot to find how consistent students are in studying and grades')
plt.show()
```
