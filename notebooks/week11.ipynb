```python
import urllib.request # this packager works with URLs and to download data
```

```python
text = urllib.request.urlopen("https://raw.githubusercontent.com/MarcusPuust/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/poem.txt").read().decode("utf-8")

text[:50]


from wordcloud import WordCloud

WordCloud().generate_from_text(text).to_image() #Converted the text we importeed from the URL to a wordcloud, then to a image
```

```python
wc = WordCloud().generate("rain bring hope hope")
img = wc.to_image()
img.show()
```

Decode (utf-8) is to convert to readable text

```python
WordCloud(stopwords=["and", "then", "but", "play"], background_color="grey").generate_from_text(text).to_image()
#The stopwords eliminate those words to appear in the wordcloud
```

```python
import urllib.request
from PIL import Image
import numpy as np

text = urllib.request.urlopen("https://drive.google.com/uc?export=download&id=1-9dlbi6ZQ9hE1gyvNzI87oBQGXG7vC6m").read().decode("utf-8")

img = urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1-9dlbi6ZQ9hE1gyvNzI87oBQGXG7vC6m",
 "image1.png")

WordCloud(mask=np.array(Image.open("image1.png")), background_color="white").generate_from_text(text).to_image()
```
